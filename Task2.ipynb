```python
# task2_cloud_deployment.py

import streamlit as st


# 1. Page Configuration

st.set_page_config(
    page_title="Cloud Deployment Planning Agent",
    layout="wide"
)

st.title("Cloud Resource Deployment Planning Agent")

st.write(
    "This agent plans the sequence of operations required "
    "to successfully deploy an AI application."
)


# 2. Problem Formulation

st.header("Problem Formulation")

st.write("**Initial State:** Deployment Request Received")

st.write("**Goal State:** AI Application Deployed Successfully")

st.write("**Performance Measures:**")
st.write("- Minimize deployment time")
st.write("- Reduce deployment cost")


# 3. Deployment Actions

actions = [
    "Allocate Resources",
    "Create Virtual Machine",
    "Install Dependencies",
    "Deploy AI Model",
    "Verify Deployment"
]

# Execution time/cost for each action
costs = {
    "Allocate Resources": 2,
    "Create Virtual Machine": 5,
    "Install Dependencies": 4,
    "Deploy AI Model": 6,
    "Verify Deployment": 2
}


# 4. Sidebar

st.sidebar.header("Deployment Planning")

application = st.sidebar.text_input(
    "Application Name",
    "My AI Application"
)

plan_button = st.sidebar.button("Generate Deployment Plan")


# 5. State Representation

if plan_button:

    st.header("Deployment State Space")

    # Initial state
    initial_state = "Deployment Request Received"

    # Generate states
    states = [
        initial_state,
        "Resources Allocated",
        "Virtual Machine Created",
        "Dependencies Installed",
        "AI Model Deployed",
        "Deployment Completed"
    ]

    # Display states
    for i, state in enumerate(states):
        st.write(f"**State {i}:** {state}")


    # 6. Execution Sequence

    st.header("Execution Sequence")

    total_cost = 0

    for i, action in enumerate(actions, start=1):

        action_cost = costs[action]
        total_cost += action_cost

        st.write(
            f"**Step {i}:** {action} "
            f"— Execution Time: {action_cost} units"
        )


    # 7. Deployment Result

    st.header("Deployment Result")

    st.success(
        f"{application} has been successfully deployed!"
    )

    st.write("**Goal Test:** Deployment status = Completed")

    st.write(
        f"**Total Deployment Cost:** {total_cost} time units"
    )


else:

    st.info(
        "Enter an application name and click "
        "'Generate Deployment Plan' to start."
    )
```
